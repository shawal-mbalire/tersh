## Interactive Plotly

Drag to rotate, scroll to zoom:

```python
import numpy as np
import plotly.graph_objects as go

x = y = np.linspace(-7.5, 7.5, 31)
r = np.hypot(*np.meshgrid(x, y))
go.Figure(go.Surface(x=x, y=y, z=np.sin(r) / r, colorscale="Viridis"))
```
